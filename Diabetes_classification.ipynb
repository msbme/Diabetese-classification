```
# Diabetes classification: SVM (RBF) & GaussianNB, with and without PCA (8 -> 4)
# Single Google Colab cell. Save as a .py or paste into one notebook cell and run.
# Outputs:
#  - Clear table of original 8-feature dataset (first 10 rows)
#  - Clear table of PCA-reduced 4 features (first 10 rows)
#  - Per-fold accuracy plots: SVM with/without PCA and NB with/without PCA
#  - ROC curves and AUC values (out-of-fold)
#  - Confusion matrices, classification reports, and aggregated evaluation metrics:
#    accuracy, precision, recall, F1-score, ROC-AUC
# Notes:
#  - If 'diabetes.csv' exists in the working directory, the script uses it.
#  - Otherwise, it prompts for upload (Colab supported).
#  - PCA is fitted inside each CV fold to avoid leakage for evaluation; a global PCA is shown only for the example table.

import os
import io
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, roc_auc_score, roc_curve, classification_report
)

# ---------------------------
# 1) Load data (local file or upload)
# ---------------------------
print("Loading dataset: trying './diabetes.csv' first...")
if os.path.exists("diabetes.csv"):
    df = pd.read_csv("diabetes.csv")
    print("Loaded './diabetes.csv'.")
else:
    try:
        # Attempt Colab upload
        from google.colab import files  # type: ignore
        print("No local file found. Please upload 'diabetes.csv' if you have it (upload dialog will appear).")
        uploaded = files.upload()
        if not uploaded:
            raise FileNotFoundError("No file uploaded.")
        # pick first file
        fname = next(iter(uploaded))
        df = pd.read_csv(io.BytesIO(uploaded[fname]))
        print(f"Loaded uploaded file: {fname}")
    except Exception as e:
        # Fallback: download canonical Pima dataset
        print("Upload not available or cancelled; downloading Pima dataset as fallback.")
        df = pd.read_csv("https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv", header=None)
        df.columns = ['Pregnancies','Glucose','BloodPressure','SkinThickness','Insulin','BMI','DiabetesPedigreeFunction','Age','Outcome']
        print("Downloaded fallback Pima dataset.")

# Ensure expected column names
expected = ['Pregnancies','Glucose','BloodPressure','SkinThickness','Insulin','BMI','DiabetesPedigreeFunction','Age','Outcome']
if list(df.columns[:9]) == expected:
    pass
else:
    # If headers present, try to detect Outcome or last column as target
    if 'Outcome' in df.columns:
        # keep as-is
        pass
    else:
        # If columns number == 9 and unnamed, set expected names
        if df.shape[1] == 9 and df.columns.dtype == object and all(str(c).startswith("Unnamed") or isinstance(c,int) for c in df.columns):
            df.columns = expected
        else:
            # If last column is target, ensure names for features
            if df.shape[1] >= 2:
                col_names = list(df.columns)
                # if last column not named Outcome, rename last to Outcome
                if col_names[-1] != 'Outcome':
                    col_names[-1] = 'Outcome'
                    df.columns = col_names

# Show dataset shape and first rows
print(f"\nDataset shape: {df.shape}")
print("\nFirst 10 rows of the original dataset (8 features + Outcome):")
display(df.head(10))

# ---------------------------
# 2) Basic cleaning / imputation (Pima common practice)
#    Replace 0 values in some physiologic measurements with NaN and impute median
# ---------------------------
df_clean = df.copy()
# Columns where zero is physiologically invalid and often used as missing in Pima dataset
zero_as_nan_cols = ['Glucose','BloodPressure','SkinThickness','Insulin','BMI']
for c in zero_as_nan_cols:
    if c in df_clean.columns:
        df_clean[c] = df_clean[c].replace(0, np.nan)

# Report missing counts
missing_counts = df_clean.isna().sum()
print("\nMissing values after zero->NaN conversion (per column):")
print(missing_counts[missing_counts > 0])

# Impute NaN with median of column
for c in zero_as_nan_cols:
    if c in df_clean.columns:
        med = df_clean[c].median()
        df_clean[c] = df_clean[c].fillna(med)

# ---------------------------
# 3) Split X / y and standardize
# ---------------------------
feature_cols = ['Pregnancies','Glucose','BloodPressure','SkinThickness','Insulin','BMI','DiabetesPedigreeFunction','Age']
X_df = df_clean[feature_cols].copy()
y = df_clean['Outcome'].astype(int).values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_df.values)

print(f"\nAfter imputation and scaling, features shape: {X_scaled.shape}")

# Show a neat table of the 8 scaled features for the first 10 rows (unscaled original for interpretability as well)
print("\nOriginal 8 features (first 10 rows, original scale):")
display(X_df.head(10))

# ---------------------------
# 4) Global PCA for demonstration table (but evaluation PCA will be fitted inside each CV fold)
# ---------------------------
pca_global = PCA(n_components=4, random_state=42)
X_pca_global = pca_global.fit_transform(X_scaled)
pca_cols = [f'PC{i+1}' for i in range(4)]
df_pca_global = pd.DataFrame(X_pca_global, columns=pca_cols)
print("\nPCA (global fit) - 4 components (first 10 rows) — NOTE: for evaluation PCA is fitted per-fold to avoid leakage:")
display(df_pca_global.head(10))

print("\nPCA explained variance ratio (global fit):")
for i, r in enumerate(pca_global.explained_variance_ratio_, 1):
    print(f"  PC{i}: {r:.4f}")
print(f"  Sum (4 PCs): {pca_global.explained_variance_ratio_.sum():.4f}")

# ---------------------------
# 5) Cross-validated evaluation (Stratified K-Fold) with out-of-fold predictions
#    We'll collect per-fold metrics and also out-of-fold probabilities to compute ROC/AUC and overall confusion matrix.
# ---------------------------
n_splits = 10
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

# Models
svm_base = SVC(kernel='rbf', C=1.0, gamma='scale', probability=True, random_state=42)
# NB doesn't provide decision_function, but has predict_proba
nb_base = GaussianNB()

# Arrays to hold out-of-fold predictions/probabilities
n_samples = X_scaled.shape[0]
oof_pred = {
    'svm_no_pca': np.zeros(n_samples, dtype=int),
    'svm_pca': np.zeros(n_samples, dtype=int),
    'nb_no_pca': np.zeros(n_samples, dtype=int),
    'nb_pca': np.zeros(n_samples, dtype=int),
}
oof_proba = {
    'svm_no_pca': np.zeros(n_samples),
    'svm_pca': np.zeros(n_samples),
    'nb_no_pca': np.zeros(n_samples),
    'nb_pca': np.zeros(n_samples),
}

# Per-fold accuracy lists for plotting
acc = {
    'svm_no_pca': [], 'svm_pca': [],
    'nb_no_pca': [], 'nb_pca': []
}

# Run CV
for fold, (train_idx, test_idx) in enumerate(skf.split(X_scaled, y), start=1):
    X_train, X_test = X_scaled[train_idx], X_scaled[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    # ---- Without PCA ----
    svm = SVC(kernel='rbf', C=1.0, gamma='scale', probability=True, random_state=42)
    svm.fit(X_train, y_train)
    pred_svm = svm.predict(X_test)
    proba_svm = svm.predict_proba(X_test)[:, 1]

    nb = GaussianNB()
    nb.fit(X_train, y_train)
    pred_nb = nb.predict(X_test)
    proba_nb = nb.predict_proba(X_test)[:, 1]

    # Store OOF
    oof_pred['svm_no_pca'][test_idx] = pred_svm
    oof_proba['svm_no_pca'][test_idx] = proba_svm
    oof_pred['nb_no_pca'][test_idx] = pred_nb
    oof_proba['nb_no_pca'][test_idx] = proba_nb

    acc['svm_no_pca'].append(accuracy_score(y_test, pred_svm))
    acc['nb_no_pca'].append(accuracy_score(y_test, pred_nb))

    # ---- With PCA (fit on train) ----
    pca = PCA(n_components=4, random_state=42)
    X_train_pca = pca.fit_transform(X_train)
    X_test_pca = pca.transform(X_test)

    svm_p = SVC(kernel='rbf', C=1.0, gamma='scale', probability=True, random_state=42)
    svm_p.fit(X_train_pca, y_train)
    pred_svm_p = svm_p.predict(X_test_pca)
    proba_svm_p = svm_p.predict_proba(X_test_pca)[:, 1]

    nb_p = GaussianNB()
    nb_p.fit(X_train_pca, y_train)
    pred_nb_p = nb_p.predict(X_test_pca)
    proba_nb_p = nb_p.predict_proba(X_test_pca)[:, 1]

    # Store OOF
    oof_pred['svm_pca'][test_idx] = pred_svm_p
    oof_proba['svm_pca'][test_idx] = proba_svm_p
    oof_pred['nb_pca'][test_idx] = pred_nb_p
    oof_proba['nb_pca'][test_idx] = proba_nb_p

    acc['svm_pca'].append(accuracy_score(y_test, pred_svm_p))
    acc['nb_pca'].append(accuracy_score(y_test, pred_nb_p))

    print(f"Completed fold {fold}/{n_splits}")

print("\nCross-validation finished.")

# ---------------------------
# 6) Compute aggregated metrics and display results
# ---------------------------
def compute_metrics(y_true, y_pred, y_proba):
    metrics = {}
    metrics['accuracy'] = accuracy_score(y_true, y_pred)
    metrics['precision'] = precision_score(y_true, y_pred, zero_division=0)
    metrics['recall'] = recall_score(y_true, y_pred, zero_division=0)
    metrics['f1'] = f1_score(y_true, y_pred, zero_division=0)
    try:
        metrics['roc_auc'] = roc_auc_score(y_true, y_proba)
    except Exception:
        metrics['roc_auc'] = np.nan
    return metrics

results = {}
for key in oof_pred.keys():
    res = compute_metrics(y, oof_pred[key], oof_proba[key])
    results[key] = res

# Present results in a tidy DataFrame
summary_rows = []
for k, v in results.items():
    summary_rows.append({
        'Config': k,
        'Accuracy': v['accuracy'],
        'Precision': v['precision'],
        'Recall': v['recall'],
        'F1': v['f1'],
        'ROC_AUC': v['roc_auc'],
        'Mean CV accuracy': np.mean(acc[k]),
        'Std CV accuracy': np.std(acc[k])
    })
summary_df = pd.DataFrame(summary_rows).sort_values('Config').reset_index(drop=True)

print("\nSummary of aggregated (out-of-fold) metrics for each configuration:")
display(summary_df.style.format({
    'Accuracy': '{:.4f}', 'Precision': '{:.4f}', 'Recall': '{:.4f}',
    'F1': '{:.4f}', 'ROC_AUC': '{:.4f}', 'Mean CV accuracy': '{:.4f}', 'Std CV accuracy': '{:.4f}'
}))

# Detailed classification reports
for k in ['svm_no_pca', 'svm_pca', 'nb_no_pca', 'nb_pca']:
    print(f"\nClassification report for {k}:")
    print(classification_report(y, oof_pred[k], digits=4))

# Confusion matrices (overall/out-of-fold)
plt.figure(figsize=(12, 10))
for i, k in enumerate(['svm_no_pca', 'svm_pca', 'nb_no_pca', 'nb_pca'], start=1):
    cm = confusion_matrix(y, oof_pred[k])
    plt.subplot(2, 2, i)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
    plt.title(f'Confusion Matrix: {k}')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
plt.tight_layout()
plt.show()

# ---------------------------
# 7) Accuracy per-fold plots (requested)
# ---------------------------
folds = np.arange(1, n_splits + 1)

plt.figure(figsize=(10, 5))
plt.plot(folds, acc['svm_no_pca'], marker='o', label=f'SVM w/o PCA (mean={np.mean(acc["svm_no_pca"]):.3f})')
plt.plot(folds, acc['svm_pca'], marker='s', label=f'SVM with PCA (4 comps) (mean={np.mean(acc["svm_pca"]):.3f})')
plt.xlabel('CV fold')
plt.ylabel('Accuracy')
plt.title('SVM (RBF): Accuracy per fold — with PCA vs without PCA')
plt.xticks(folds)
plt.ylim(0.4, 1.02)
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 5))
plt.plot(folds, acc['nb_no_pca'], marker='o', label=f'NB w/o PCA (mean={np.mean(acc["nb_no_pca"]):.3f})')
plt.plot(folds, acc['nb_pca'], marker='s', label=f'NB with PCA (4 comps) (mean={np.mean(acc["nb_pca"]):.3f})')
plt.xlabel('CV fold')
plt.ylabel('Accuracy')
plt.title('GaussianNB: Accuracy per fold — with PCA vs without PCA')
plt.xticks(folds)
plt.ylim(0.4, 1.02)
plt.legend()
plt.grid(True)
plt.show()

# ---------------------------
# 8) ROC curves (out-of-fold probabilities) and AUCs
#    We'll plot all four curves together for comparison.
# ---------------------------
plt.figure(figsize=(8, 8))
for label, color in zip(['svm_no_pca','svm_pca','nb_no_pca','nb_pca'], ['C0','C1','C2','C3']):
    try:
        fpr, tpr, _ = roc_curve(y, oof_proba[label])
        auc = roc_auc_score(y, oof_proba[label])
        plt.plot(fpr, tpr, lw=2, label=f"{label} (AUC={auc:.3f})", color=color)
    except Exception as e:
        print(f"Could not compute ROC for {label}: {e}")
plt.plot([0,1],[0,1],'k--', alpha=0.5)
plt.xlim([-0.01,1.01])
plt.ylim([-0.01,1.01])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC curves (out-of-fold) for all configurations')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()

# ---------------------------
# 9) PCA effects: show explained variance and compare classification on principal components
# ---------------------------
print("\nPCA (global) explained variance ratio (4 components):")
for i, r in enumerate(pca_global.explained_variance_ratio_, 1):
    print(f"  PC{i}: {r:.4f}")
print(f"  Sum explained by 4 PCs: {pca_global.explained_variance_ratio_.sum():.4f}")

# Show example of how the first 10 rows map from 8 original features to 4 PCA components (global PCA used only for demonstration)
display(pd.concat([X_df.head(10).reset_index(drop=True), df_pca_global.head(10)], axis=1))

# ---------------------------
# 10) Final comparison summary (formatted)
# ---------------------------
print("\nFinal comparison (mean CV accuracy ± std and overall ROC AUC):")
comp_rows = []
for key in ['svm_no_pca','svm_pca','nb_no_pca','nb_pca']:
    comp_rows.append({
        'Config': key,
        'Mean CV Accuracy': np.mean(acc[key]),
        'Std CV Accuracy': np.std(acc[key]),
        'ROC_AUC (oof)': results[key]['roc_auc']
    })
comp_df = pd.DataFrame(comp_rows).sort_values('Config').reset_index(drop=True)
display(comp_df.style.format({'Mean CV Accuracy': '{:.4f}', 'Std CV Accuracy': '{:.4f}', 'ROC_AUC (oof)': '{:.4f}'}))

print("\nDone. The plots above include the requested accuracy comparisons and ROC/AUC. Classification reports and confusion matrices are shown for each configuration.")
```
